##Introducció


**Google Colaboratory**, conegut com a **Google Colab**, és un servei de Google que permet crear i executar Jupyter Notebooks dins del Google Drive en l'entorn ade Google Workspace.
D'aquesta manera no hi ha la necessitat de tenir intal·lar Python ni configurar un entorn de desenvolupament al nostre ordinador.


## Què és un Notebook?

Podem imaginar un **Notebook** com una **llibreta digital** de laboratori. En una llibreta tradicional podríem escriure què volem experimentar, fer els càlculs i apuntar-ne els resultats. En un Notebook podem fer exactament el mateix, però amb l'avantatge que el codi Python es pot executar directament.

Quan treballem amb Colab, el nostre navegador actua com a interfície, però el codi normalment no s'executa al nostre ordinador, sinó en una màquina virtual proporcionada per Google.

Un Notebook no és un únic fitxer que s'executa necessàriament de principi a final. Està dividit en petits blocs anomenats **cel·les**.

Cada cel·la pot ser de dos tipus:

* [1. Cel·les de codi](#1-celles-de-codi)
* [2. Cel·les de text](#2-celles-de-text)

###1. Cel·les de codi
Les cel·les de codi contenen instruccions d'un llenguatge concret, en el nostre cas serà de Python.
Quan executem una cel·la de codi, Colab envia el codi a un entorn d'execució de Python i mostra el resultat just a sota.

Podem executar una cel·la utlitzant el botó especificat o mitjançant la combinació ***Shift + Enter***

**Les variables creades en una cel·la poden ser utilitzades posteriorment en altres cel·les**

Per exemple:



````python
# EXEMPLE 1
# Declarem i assignem valors a dues variables (preu i iva)
preu = 150
iva = 0.21
````

La cel·la posterior coneix les variables `preu` i `iva` perquè la primera cel·la ja s'ha executat

````python
# Utilitzem les variables per crear-ne una de nova i mostrar el resultat
preu final = preu + preu * iva
print(preu final)
````

Si tenim diferents cel·les de codi en el Notebook el més habitual és executar-les de forma seqüencial, però una de les avantatges que tenim, però que també hem d'anar més en compte, és que podem modificar cel·les concretes i executar només aquella cel·la sense prèviament executar les anteriors.
Compte! que l'entorn de variable serà del que disposem si no hem executat les cel·les anteriors.

En l'exemple anterior podríem modificar el contingut de la cel·la anterior i només executar aquesta sense haver de tornar a executar la primera cel·la si les variables a utilitzar i els seus valors inicials ja són els correctes.


###2. Cel·les de text

Les cel·les de text serveix per documentar el Notebook.
No contenen codi Python executable i les podem utilitzar per explicar què estem fent, per què ho fem, quines decisions hem près, quins resultats o conclusions hem obtingut i podem extreure.

La sintàxi d'una cel·la de text segueix l'estructura i llenguatge de marques anomenat [Markdown](https://www.markdownguide.org/) molt utilitzat sobetot en repositoris de GitHub.

##Entorn d'execució Google Colab

Quan obrirm un Notebook de Google Colab, Google ens proporciona temporalment una **màquina virtual** (sandbox) on s'executarà el nostre codi.

Aquesta màquina dipsosa de CPU, RAM, Emmagatzematge temporal, Python i diverses llibreries preinstal·lades.

Podem canviar l'Entorn d'execució a través del menú:
> *Entorn d'execució* -> *Canviar Tipus Entorn Execució*

Des d'allà serà possible executar el nostre Notebook en una **CPU**, **GPU** o en una **TPU** de manera gratuita amb certes restricions. Per exemple que la **sessió només dura 12hores**.

### Accés al sistema de fitxers del Drive

Podem accedir a l'estructura de Directoris del nostre Google Drive mitjnçant la llibreria `google.colab` i important la classe `drive`.

La primera vegada que executem `drive.mount('/content/drive')` Google ens demanarà permís perquè Colab tingui accés.

````python
import os
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')


path = '/content/drive/'
#path = '/content/drive/MyDrive/Colab Notebooks/M0572-SistemesAprenentatgeAutomatic/01_introduccio/'

#Canviem de directori per el que hi hagi a la variable path
os.chdir(path)

print(os.getcwd())
````

Podem també accedir al sistema de fitxers mitjançant el desplegable dret i obrin la icona de la carpeta 📁. Allà veuràs l'estructura de directoris del Drive d'una forma més visual a on també hi podràs pujar fitxers (upload).

##Cheatsheet bàsic de Google Colab / Jupyter

| Acció                                              | Drecera a Google Colab  |
| -------------------------------------------------- | ----------------------- |
| Executar cel·la i anar a la següent             | `Shift + Enter`         |
| Executar cel·la sense avançar                   | `Ctrl + Enter`          |
| Executar i inserir una nova cel·la de codi       | `Alt + Enter`           |
| Nova cel·la de **codi a sota**                   | `Ctrl + M`, després `B` |
| Nova cel·la de **codi a sobre**                  | `Ctrl + M`, després `A` |
| Convertir la cel·la actual en **text/Markdown** | `Ctrl + M`, després `M` |
| Convertir la cel·la actual en **codi**          | `Ctrl + M`, després `Y` |
| Esborrar la cel·la actual                      | `Ctrl + M`, després `D` |
| Desfer l'eliminació d'una cel·la                | `Ctrl + M`, després `Z` |
| Tallar cel·la                                   | `Ctrl + M`, després `X` |
| Copiar cel·la                                   | `Ctrl + M`, després `C` |
| Enganxar cel·la                                 | `Ctrl + M`, després `V` |
| Moure cel·la cap amunt                          | `Ctrl + M`, després `K` |
| Moure cel·la cap avall                          | `Ctrl + M`, després `J` |
| Comentari Python                                | `# comentari`           |
| Ajuda d'una funció                              | `help(funció)`          |
| Executar comanda Linux                          | `!comanda`              |
| Instal·lar paquet                               | `!pip install paquet`   |
| Llistar fitxers del directori                   | `!ls`                   |
| Directori actual                                | `%pwd`                  |
| Canviar directori                               | `%cd carpeta`           |
| Mesurar temps (quan tarda una instrucció)       | `%time expressió`       |
| Mesurar diverses execucions                     | `%timeit expressió`     |
| Variables existents                             | `%whos`                 |
