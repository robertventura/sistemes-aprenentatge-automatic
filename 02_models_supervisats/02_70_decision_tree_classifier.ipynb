## 🎯 Objectius
- Aquest notebook pretén mostrar un exemple senzill d'ús d'un model d'arbre de desició en l'àmbit de la classifiació
- Hi ha un exemple utilitzant Gini Índex i Croos-Entropy com a mesura d'impuresa


## 🟦 1. Gini Índex

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
import matplotlib.pyplot as plt

# ============================
# 1. Definir el dataset
# ============================

data = {
    "X1": [2, 3, 3, 3, 5, 6, 7, 7],
    "Classe": ["A", "A", "A", "A", "B", "A", "B", "B"]
}

df = pd.DataFrame(data)

X = df[["X1"]]
y = df["Classe"]

# ============================
# 2. Entrenar el model
# ============================

model = DecisionTreeClassifier(
    criterion="gini",  # utilitza cross entropy
    random_state=42,
    max_depth=None        # deixem créixer l’arbre al màxim
)
model.fit(X, y)


# ============================
# 4. Dibuixar l’arbre
# ============================

plt.figure(figsize=(10, 6))
plot_tree(
    model,
    feature_names=["X1"],
    class_names=model.classes_,
    filled=True,
    rounded=True,
    impurity=True,
    fontsize=12
)
plt.title("DecisionTreeClassifier – Gini Índex")
plt.show()


```

## 🟧 2. Cross Entropy

```python
import pandas as pd
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
import matplotlib.pyplot as plt

# ============================
# 1. Definir el dataset
# ============================

data = {
    "X1": [2, 3, 3, 3, 5, 6, 7, 7],
    "Classe": ["A", "A", "A", "A", "B", "A", "B", "B"]
}

df = pd.DataFrame(data)

X = df[["X1"]]
y = df["Classe"]

# ============================
# 2. Entrenar el model
# ============================

model = DecisionTreeClassifier(
    criterion="entropy",  # utilitza cross entropy
    random_state=42,
    max_depth=None        # deixem créixer l’arbre al màxim
)
model.fit(X, y)


# ============================
# 4. Dibuixar l’arbre
# ============================

print(tree.export_text(model))

plt.figure(figsize=(10, 6))
plot_tree(
    model,
    feature_names=["X1"],
    class_names=model.classes_,
    filled=True,
    rounded=True,
    impurity=True,
    fontsize=12
)
plt.title("DecisionTreeClassifier – Cross Entropy")
plt.show()


```
